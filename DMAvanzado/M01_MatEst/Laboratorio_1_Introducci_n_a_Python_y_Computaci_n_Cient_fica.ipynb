### Tipos de datos básicos

`int`: Números enteros
- Por ejemplo: `1`, `10`, `-4`, ...

`float`: Números decimales
- Por ejemplo: `3.14`, `-1.25`, `1.0`, `0.0`, ...
- Notar que todo `int` puede ser representado como `float`, pero no todo `float` puede representarse como un `int`.

`str`: Cadenas de caracteres (texto)
- Por ejemplo: `"hola"`, `"¡Buenos días!"`, `'1234'`, ...
- Notar que todo `int` o `float` puede ser representado como `str`. Los `str` pueden estar entre "comillas dobles" o 'comillas simples'.

`bool`: Verdadero o falso
- Sólo pueden ser `True` o `False`.

Hay más tipos de datos, que iremos viendo a medida que los usemos.

````python
# Operaciones interesantes con números (int o float)
print(5 + 3)   # Suma
print(5 - 3)   # Resta
print(5 * 3)   # Multiplicación
print(10 / 3)  # División
print(10 // 3) # División entera (trunca el resultado al entero menor más cercano)
print(10 % 3)  # Módulo (resto de una división)
print(10 ** 3)  # Potencia
````

````python
# Operaciones interesantes con strings (str)

saludo = "Hola"
apellido = "García"
print(len(saludo))     # Largo
print(f"{saludo} señor {apellido}, cómo está?")  # Concatenar
print(saludo.upper())  # Mayúsculas
print(saludo.lower())  # Minúsculas
````

````python
# Operaciones interesantes con booleanos (bool)

a = True
b = False
print(a)
print(b)
print(type(a)) # Imprimir de qué tipo es la variable a

print() # Imprimir línea en blanco

# Operadores lógicos
print(True and True)  # True
print(True and False) # False
print(False or True)  # True
print(False or False) # False
print(not True)       # False
print(not False)      # True

print() # Imprimir línea en blanco

# Comparaciones
print(5 > 3)  # True
print(5 < 3)  # False
print(5 == 5) # ¿Son iguales? True
print(5 != 3) # ¿Son distintos? True
print(5 >= 5) # True
print(4 <= 3) # False
````

````python
# Sumar strings con números
a = "Mi número favorito es: "
b = 24
print(a + b) # Esto lanzará un error! No se pueden sumar strings con números
````

### Estructuras de datos básicas

`list`: Listas
- Por ejemplo,
  - `[1, 2, 3]`
  - `['Pedro', 'Juan', 'Diego']`
  - `[1, 'a', 2.0]` (Pueden tener elementos de varios tipos)
  - `[["a", "b"], 1, 2]` (Puede haber una lista dentro de otra)
  - ...

`tuple`: Tuplas
- Son como listas, pero que luego no se pueden modificar. Son más rápidas de procesar.
- Por ejemplo,
  - `('a', 'b')`
  - `(1, 2, 3)`
  - ...

`dict`: Diccionarios
- Guarda datos en pares `clave: valor`.
- Por ejemplo:
  - `persona = {"nombre": "Ana", "edad": 20, "genero": "femenino"}`


````python
# Operaciones útiles con listas

equipos = ["Camión", "Perforadora", "Pala"]

equipos.append("Chancador")     # agregar
equipos[0] = "Camión CAEX"      # modificar el primer elemento (se empieza a contar desde el 0)
equipos.remove("Perforadora")   # eliminar
print(equipos)                  # ["Camión CAEX", "Pala", "Chancador"]
print(len(equipos))             # Obtener el largo de la lista
````

````python
# Operaciones útiles con tuplas

equipos = ["Camión", "Perforadora", "Pala"]
print(equipos[1])         # Obtener el elemento en la posición 1 (se empieza a contar desde el 0)
print(len(equipos))       # Obtener el largo de la tupla
````

````python
# Operaciones útiles con diccionarios

alumno = {
    "nombre": "Annie",
    "edad": 20,
    "carrera": "Ingeniería Civil de Minas",
    "aprobado": True
}

# Leer un valor
print(alumno["nombre"])    # Ana
print(alumno["edad"])      # 20

# Modificar un valor
alumno["edad"] = 21
print(alumno["edad"])      # 21

# Agregar una clave nueva
alumno["promedio"] = 6.2
print(alumno["promedio"])  # 6.2

# Eliminar una clave
del alumno["aprobado"]

# Verificar si una clave existe
print("nombre" in alumno)   # True
print("rut" in alumno)      # False
````

### Condicionales

Nos permiten hacer que dada una condición se ejecuten (o no) ciertas líneas del código.

````python
if 1 + 1 == 2: # Si se cumple la condición,
  print("Esto se ejecutará") # Se ejecuta esta línea

if 1 + 1 == 3: # Esta condición no se cumple
  print("Esto no se ejecutará") # Entonces esta línea no se ejecuta

print("Y esta línea se ejecutará?") # Sí, porque está fuera del if
````

````python
nota = 4.1

if nota >= 4.0: # Si se cumple esta condición,
  print("Aprobado") # se ejecutará esta línea.
else: # Si no se cumple la condición,
  print("Reprobado") # se ejecuta esta otra línea.
````

### Bucles

Al hacer una tarea repetitiva, se busca evitar escribir muchas veces un mismo código. Para eso, existen los bucles.

````python
# Bucle for: Recorre los elementos de alguna estructura

# La función range(x, y) genera una secuencia de numeros desde x hasta y-1
for i in range(0, 5): # Para cada número del 0 al 4...
  print(i)

print() # Línea en blanco

alturaDeVolcanes = {"Tupungato": 6570, "Pular": 6233, "Ojos del Salado": 6893} # Esto es un diccionario
for volcan in alturaDeVolcanes: # Para cada elemento del diccionario...
  print(f"El {volcan} mide {alturaDeVolcanes[volcan]} metros.")
````

````python
# Bucle while: Repite el código mientras la condición sea True

contador = 0
while contador < 5: # Mientras el contador sea menor a 5, repetir lo que viene a continuación
    print(contador)
    contador += 1 # Sumar uno al contador (equivalente a decir contador = contador + 1)

print()

i = 0
while True: # Mientras True sea True (siempre!) repetir lo que viene a continuación
    i += 1
    print(i)
    if i == 3:
        break # Salir del loop forzosamente
````

### Funciones

Reciben ciertos parámetros, y hacen algún trabajo según como se definan. Son piezas fundamentales a medida que los programas se van haciendo más complejos.


La estructura general de una función:

```python
def nombre_funcion(parámetro_1, parámetro_2, ..., parámetro_n):
  
  # Aquí iría el algoritmo de la función...
  ...

  # Retornar el resultado de la función
  return resultado
```

Por ejemplo, una función útil puede ser:

````python
# Función básica
def saludar():
    print("Hola, mundo")

saludar() # Ejecuta todo lo que diga def saludar()
````

````python
# Con parámetros
def saludar(nombre):
    print(f"Hola, ¡{nombre}!")

saludar("Annie")  # Hola, ¡Annie!
saludar("Erwin") # Hola, Erwin!
````

````python
# Que retorne algo
def sumar(numeros):

  acumulado = 0 # Inicialmente la suma es cero
  for n in numeros: # Para cada número de la lista
    acumulado += n # Se suma n al acumulado

  return acumulado # Al final retornamos la sumatoria de todos los números

resultado = sumar([1, 2, 4])
print(resultado) # 1 + 2 + 4 = 7
````

````python
# Con parámetros por defecto
def saludar(nombre, saludo="Hola"):
    print(f"{saludo}, ¡{nombre}!")

saludar("Annie")           # → Hola, Ana!
saludar("Annie", "Buenas") # → Buenas, Ana!
````

````python
# Con condicional dentro
def clasificar_nota(nota):
    if nota >= 4.0:
        return "Aprobado"
    else:
        return "Reprobado"

print(clasificar_nota(6.5)) # Aprobado
print(clasificar_nota(2.0)) # Reprobado
````

### Resumen de los operadores


##### Aritméticos

| Símbolo | Tarea ejecutada | Ejemplo |
|:---:|---|---|
| `+` | Suma | `13+2` → 15
| `-` | Resta | `13-2` → 11
| `*` | Multiplicación | `13*2` → 26
| `**` | Potencia | `13**2` → 169
| `/` | División | `13/2` → 6.5
| `//` | División entera / Piso | `13//2` → 6
| `%` | Módulo | `13%2` → 1



##### Relacionales

|   Símbolo   | Tarea ejecutada | Ejemplo
|:---:|---|---|
| `==` | igual | `10==2` → False
| `!=`  | no igual | `10!=2` → True
| `<` | menor que | `10<2` → False
| `>` | mayor que | `10>2` → True
| `<=`  | menor o igual que | `10<=2` → False
| `>=`  | mayor o igual que | `10>=2` → True

##### Lógica

| Símbolo | Logica | Ejemplo
|:---:|---|---|
| `and` | y | `True and False` → False |
| `or` | o | `True or False` → True |
| `not` | negación | `not (True or False)` → False |

##### Pertenencia

| Símbolo | Tarea ejecutada | Ejemplo
|:---:|:---|:---
| `in` | Verifica si un objeto está contenido en otro | `for numero in [1, 2, 3]:` ...
